```python
# -*- coding: utf-8 -*-
"""
TP_1

@author: Rath
"""
# Importing the libraries
import pandas as pd
import matplotlib.pyplot as plt
#Import scikit-learn metrics module for accuracy calculation
from sklearn import metrics
#Import knearest neighbors Classifier model
from sklearn.neighbors import KNeighborsClassifier

# Importing the dataset
dataset = pd.read_csv('base3.csv')


##KNN without normalization

print("\n\n 4 \n\n")

##KNN with normalization with addition of Sex variable

## transform qualitative variable in binary in dataset
from pandas import get_dummies
dataset =get_dummies(dataset)

# Prepocessing data
from sklearn.preprocessing import MinMaxScaler
scaler=MinMaxScaler(feature_range=(0, 1)) 
dataset['Age']=scaler.fit_transform(dataset[['Age']])
dataset['EstimatedSalary']=scaler.fit_transform(dataset[['EstimatedSalary']])

### Extract input and outup data
X2 = dataset.iloc[:,[1,2,4,5]].values
y2 = dataset.iloc[:, -3].values

# Splitting the dataset into the Training set and Test set
from sklearn.model_selection import train_test_split
X2_train, X2_test, y2_train, y2_test = train_test_split(X2, y2, test_size = 0.3, random_state = 1)

# Let fix k to a chosen number
fixed_k = 100
accuracy_set = []
for i in range(1, fixed_k):
    #Create KNN Classifier
    knn = KNeighborsClassifier(n_neighbors=i)
    
    #Train the model using the training sets
    knn.fit(X2_train, y2_train)
    if i == 10:
        print(knn.n_neighbors)
        print(knn.fit)
        print(knn.kneighbors().count())
        print(knn.kneighbors())
        print(dir(knn))
    
    #Predict the response for test dataset
    y2_predc = knn.predict(X2_test)
    # Model Accuracy, how often is the classifier correct?
    accuracy_set.append((i, metrics.accuracy_score(y2_test, y2_predc)))
# print("Accuracy:", metrics.accuracy_score(y2_test, y2_predc))

# Tracer le graphique du coude pour déterminer le meilleur k
plt.figure(figsize=(10, 6))
plt.plot(range(1, fixed_k), list(map(lambda x: x[1], accuracy_set)))
plt.title('Le graphe de précision en fonction des k')
plt.ylabel('Précision')
plt.xlabel('K')
plt.show()

print(f"\n\n Le meilleur K est: {sorted(accuracy_set, key=lambda x: x[1], reverse=True)[0]} \n\n")

## compute the confusion matrix
# from sklearn.metrics import confusion_matrix
# _cmatrix = confusion_matrix(y2_test, y2_predc)
# print(_cmatrix)
```
